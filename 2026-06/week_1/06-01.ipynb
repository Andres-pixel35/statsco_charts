# Dataset description
**Data Source**: PISA Performance Scores by Country by Dennis Kao

**Original Author**: [Data Source](https://data.world/professorkao)

**Download**: [Kaggle](https://www.kaggle.com/datasets/thedevastator/pisa-performance-scores-by-country/data)

## Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Code

```python
path = "./data/OECD PISA data.csv"

df = pd.read_csv(path)

col_df = df[df["LOCATION"] == "COL"]
col_df = col_df.drop(columns=["index"])
col_df = col_df[col_df["TIME"] == 2018]
a = col_df.groupby(["INDICATOR", "SUBJECT"])["Value"].sum()
```

## CHART

```python
subjects = ["BOY", "GIRL"]
indicators = ["PISAMATH", "PISAREAD", "PISASCIENCE"]
clean_indicator = ["MATH", "READ", "CIENCE"]
colors = {"BOY": "teal", "GIRL": "coral"}


plt.figure(figsize=(10.8, 8), dpi=300)

for subject in subjects:
    values = [a[indicator][subject] for indicator in indicators]
    plt.plot(clean_indicator, values, marker="o", color=colors[subject])
    plt.text(2.02, values[-1] - 0.2, f"{subject}", color=colors[subject])

plt.gca().spines["top"].set_visible(False)
plt.gca().spines["bottom"].set_visible(False)
plt.gca().spines["left"].set_visible(False)
plt.gca().spines["right"].set_visible(False)

plt.tick_params(axis="y", length=0)
plt.ylabel("Mean Score", labelpad=10, y=0.85)
plt.xlabel("Category", labelpad=10)

plt.show()
```
