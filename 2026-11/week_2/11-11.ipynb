# Code
## Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Load data

```python
path = "./data/maltrato_animal.csv"
path_pop = "./data/departamental.csv"

df = pd.read_csv(
    path, encoding="utf-8",
    usecols=[3,4,8, 11],
    dtype={"Código Dane": str}
)
# DANE projections; "Total" is the total population of the department
pop = pd.read_csv(path_pop, usecols=["DP", "AÑO", "Total"])
```

## Clean Data

```python
df.groupby("Departamento")["Cantidad"].sum()
```

```python
df.groupby("Clase de Sitio")["Cantidad"].sum()
```

```python
cases = df.groupby(df["Código Dane"].str[:2].astype(int))["Cantidad"].sum()
pop_2026 = pop[pop["AÑO"].eq(2026)].set_index("DP")["Total"]

rate = (
    cases.to_frame("cases")
    .join(pop_2026.rename("population"), validate="1:1")
)
rate["per_100k"] = rate["cases"] / rate["population"] * 100_000
rate["Departamento"] = df.groupby(df["Código Dane"].str[:2].astype(int))["Departamento"].first()
rate.sort_values("per_100k", ascending=False)
```
