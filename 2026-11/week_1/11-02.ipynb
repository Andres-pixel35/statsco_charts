# Dataset Description
## Top 5 countries of residence of non-resident foreigners entering Colombia, Jan-Jun 2026

**Source**: [Ministerio de Comercio, Industria y Turismo](https://www.datos.gov.co/Comercio-Industria-y-Turismo/Extranjeros-No-Residentes/7wm8-w5ad/about_data)

# Code
## Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Load data

```python
path = "./data/Extranjeros_No_Residentes_20261031.csv"

df = pd.read_csv(path, encoding="utf-8", usecols=[0,4,5], thousands=",")
```

## Clean data

```python
df_f = df[df["Año"].eq(2026)].drop(columns=["Año"])
```

```python
total = df_f.groupby("PaisOEEResidencia").sum().reset_index()
```

```python
print("Top 5:\n", total.nlargest(5, "Cant Extranjeros no Residentes"))
```
