# Dataset description
## Number of Captured

Source: [Policia Nacional](https://portalsiedco.policia.gov.co:4443/extensions/PortalPublico/index.html#/home)

# Code
## Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Load data

```python
path = "./data/capturas.csv"

df = pd.read_csv(path, encoding="utf-8")
```

## Clean data

```python
df_f = df[df["mes"].eq(8)].drop(columns=["fecha", "mes"])
```

## Plot

```python
background = "#F4F1EA"
blue = "#4a90e2"
x_axis = df_f["año"]

fig, ax = plt.subplots(figsize=(9.8, 7.0), dpi=500)
fig.patch.set_facecolor(background)
ax.patch.set_facecolor(background)

plt.bar(x_axis, df_f["capturas"], color=blue)

plt.xticks(x_axis, rotation=45)

plt.gca().yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f"{int(x/1000)}K" if x != 0 else "0")
)

for side in ["top", "bottom", "left", "right"]:
    plt.gca().spines[side].set_visible(False)

for axis in ["y", "x"]:
    plt.tick_params(axis=axis, length=0)

plt.ylabel("Captures", labelpad=10, fontsize=10, y=0.9)

plt.show()
```
