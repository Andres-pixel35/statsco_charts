# Dataset description
## Cases of terrorism per year

**Source**: MinDefensa

**Download**: [link](https://www.mindefensa.gov.co/defensa-y-seguridad/datos-y-cifras/informacion-estadistica)

# Code
## Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Load data

```python
path_terr = "./data/TERRORISMO.csv"
path_pop = "./data/departamental.csv"

cases_cols = [
    "FECHA_HECHO", "COD_DEPTO", "DEPARTAMENTO", "CANTIDAD"
]

pop_cols = [
    "DP", "DPNOM", "AÑO", "Total"
]

cases = pd.read_csv(path_terr, encoding="utf-8", usecols=cases_cols)
pop = pd.read_csv(path_pop, encoding="utf-8", usecols=pop_cols)
```

## Clean data

```python
cases["FECHA_HECHO"] = pd.to_datetime(cases["FECHA_HECHO"])
cases["FECHA_HECHO"] = cases["FECHA_HECHO"].dt.year
cases_f = cases.groupby(["FECHA_HECHO", "COD_DEPTO"]).sum(numeric_only=True).reset_index()

total = pd.merge(
    cases_f, 
    pop, 
    left_on=["FECHA_HECHO", "COD_DEPTO"],
    right_on=["AÑO", "DP"],
    how="left"
)[["FECHA_HECHO", "COD_DEPTO", "DPNOM", "CANTIDAD", "Total"]]

total["Ratio"] = (total["CANTIDAD"] / total["Total"]) * 100000

total_2026 = total[total["FECHA_HECHO"].eq(2026)]

total_2026.nlargest(n=5, columns=["Ratio"])[["DPNOM", "Ratio"]]
```
