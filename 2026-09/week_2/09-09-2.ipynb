# Code
## Imports

```python
from types import LambdaType
import pandas as pd
import matplotlib.pyplot as plt
```

## Load data

```python
path = "./data/annual_inflation_2026_categories_all_months.csv"

df = pd.read_csv(path, encoding="utf-8")
```

## Plot

```python
background = "#F4F1EA"
gray = "#D3D3D3"

category_colors = {
    "Restaurants and hotels": "#264653",
    "Health": "#E76F51",
    "Education": "#2A9D8F",
}

fig, ax = plt.subplots(figsize=(9.8, 7.0), dpi=500)
fig.patch.set_facecolor(background)
ax.patch.set_facecolor(background)

for column in df.columns[1:]:
    plt.plot(
        df["Month"],
        df[column],
        linewidth=2,
        color=category_colors.get(column, gray)
    )
    if column in category_colors.keys():
        plt.annotate(
            f"{df[column].iloc[-1]:.1f}%",
            xy=(len(df)-1 + 0.1, df[column].iloc[-1] - 0.1),
        )

for side in ["top", "bottom", "left", "right"]:
    plt.gca().spines[side].set_visible(False)

for axis in ["y", "x"]:
    plt.tick_params(axis=axis, length=0)

plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f"{x:.0f}%"))

plt.ylabel("Inflation", labelpad=10, fontsize=10, y=0.9)
plt.xlabel("Month", labelpad=10, fontsize=10)
    
plt.show()
```
